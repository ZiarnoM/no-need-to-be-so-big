## Ollama llm usage demo notebook

This notebook demonstrates how we call models run on local ollama

```python
import json
import requests
url = "http://localhost:11434/api/generate"
```

’’’
ollama serve
’’’

```python
prompt = input("Enter your prompt: ")
data = {
    "model": "qwen3.5:9b",
    "keep_alive": -1,
    "prompt": prompt,
    "stream": False,
    "temperature": 0
    }
```

```python
response = requests.post(url, json=data)
response.raise_for_status()
```

```python
result = response.json()
full_response = result.get("response", "")
print("LLM response: " + full_response)
```
